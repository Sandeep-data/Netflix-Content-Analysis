```python
# Libraries import
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Dataset read
df = pd.read_csv("netflix_titles.csv")
```

```python
# top 5 head
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# Statistics
df.describe(include="all")
```

```python
#Missing Values
df.isnull().sum()
```

```python
# Duplicate Rows
df.duplicated().sum()
```

```python
#Remove Duplicate
df = df.drop_duplicates()
```

```python
df.shape
```

```python
# Movies vs TV Shows
df["type"].value_counts()
```

```python
# Visualization
plt.figure(figsize=(6,4))
sns.countplot(x="type", data=df)
plt.title("Movies vs TV Shows")
plt.show()
```

```python
# Rating Count
plt.figure(figsize=(10,5))
sns.countplot(y="rating", data=df)
plt.show()
```

```python
# Top 10 Countries
df["country"].value_counts().head(10)
```

```python
# Graph
plt.figure(figsize=(10,5))
df["country"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries")
plt.show()
```

```python
# Top 10 Directors
df["director"].value_counts().head(10)
```

```python
# Release Year
plt.figure(figsize=(12,5))
sns.histplot(df["release_year"], bins=30)
plt.show()
```

```python
# Movies after 2020
df[df["release_year"]>=2020]
```

```python
# Comedy Movies
df[df["listed_in"].str.contains("Comedy", na=False)]
```

```python
# India Movies
df[df["country"].str.contains("India", na=False)]
```

```python
# Highest Duration
df["duration"].value_counts().head(10)
```
